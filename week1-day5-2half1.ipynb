```python
import pandas as pd

df = pd.read_csv("clean_data.csv")

print(df.head())
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt

df["department"].value_counts().plot(kind="bar")

plt.title("Department Court")
plt.xlabel("Department")
plt.ylabel("Count")

plt.show()
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
df["department"].value_counts().plot(kind="bar")

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.savefig("reports/figures/bar_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=10)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.savefig("reports/figures/histogram.png")
plt.show()
```

```python
print(df.columns)
```

```python
print(df["age"].describe())
```

```python
print(df.nlargest(10,"age")[["employee_id","name","age"]])
```

```python
df.loc[df["age"]>100,"age"]=0
```

```python
print(df["age"].describe())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"].replace(0, float("nan")).dropna(),bins=10)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("number of Employees")

plt.show
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["age"], df["salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.savefig("reports/figures/scatter_plot.png")
plt.show()
```

```python
sorted_salary = df["salary"].sort_values().reset_index(drop=True)

plt.figure(figsize=(8, 5))

plt.plot(sorted_salary, marker="o")

plt.title("Employee Salary Trend")
plt.xlabel("Employee Index")
plt.ylabel("Salary")
plt.savefig("reports/figures/line_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["salary"])

plt.title("Employee Salary Distribution")
plt.ylabel("Salary")
plt.savefig("reports/figures/box_plot.png")
plt.show()
```
